Custo carro novo

```


#Recebe o preco de fabrica
custo_fabrica = float(input('Digite o custo de fabrica do carro: R$ '))
```

```
#Calcula o valor final
impostos = custo_fabrica*0.25
valor_com_impostos = custo_fabrica+impostos
distribuicao = valor_com_impostos*0.35
custo_final = valor_com_impostos+distribuicao
```

```


#Recebe o resultado
print(f"\nCusto ao consumidor: r$ {custo_final:.2f}")
```

Conversão de horas e minutos

```
#Valores iniciais
tempo_inicial = 5
minutos_iniciais = 35
total_itens = 160
itens_processados = 3
```

```


#Calculo de taxa de processamento por minuto
taxa_processamento = itens_processados / total_itens
print(f"{itens_processados}/{total_itens} = {taxa_processamento:.2f}".replace('.', ','))
```

```


#Calculo do tempo total em horas
tempo_total = tempo_inicial + taxa_processamento # 5 + 0,51
print(f"{tempo_inicial} + {taxa_processamento:.2f} = {tempo_total:.2f}".replace('.', ','))
```

```


#Exibicao do resultado final
print(f"{tempo_total:.2f}".replace('.', '.'))
```

Hora extra

```
# Dados de entrada
tempo_total_horas = 10 + 45/60  # 10h45min = 10,75 horas
valor_hora = 25.00
```

```
# Cálculo das horas
horas_normais = 8
horas_extras = tempo_total_horas - horas_normais
```

```
# Pagamentos
pagamento_normais = horas_normais * valor_hora
valor_hora_extra = valor_hora * 1.7  # 70% a mais
pagamento_extras = horas_extras * valor_hora_extra
```

```
# Total
pagamento_total = pagamento_normais + pagamento_extras
```

```
# Resultados
print(f"João trabalhou {tempo_total_horas:.2f} horas")
print(f"8h normais → R${pagamento_normais:.2f}")
print(f"{horas_extras:.2f}h extras → R${pagamento_extras:.2f}")
print(f"Total: R${pagamento_total:.2f}")
```
