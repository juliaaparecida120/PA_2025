Conversão de horas e minutos

Hora extra

```
# Início
horas = 2
minutos = 45
```

```
# Conversão de minutos para horas decimais
minutos_em_horas = minutos / 60
```

```
# Soma total em horas decimais
total_horas = horas + minutos_em_horas
```

```
# Exibe o resultado
print("Total em horas decimais:", total_horas)
```

Hora extra

```
# Dados de entrada
tempo_total_horas = 10 + 45/60  # 10h45min = 10,75 horas
valor_hora = 25.00
```

```
# Cálculo das horas
horas_normais = 8
horas_extras = tempo_total_horas - horas_normais
```

```
# Pagamentos
pagamento_normais = horas_normais * valor_hora
valor_hora_extra = valor_hora * 1.7  # 70% a mais
pagamento_extras = horas_extras * valor_hora_extra
```

```
# Total
pagamento_total = pagamento_normais + pagamento_extras
```

```
# Resultados
print(f"João trabalhou {tempo_total_horas:.2f} horas")
print(f"8h normais → R${pagamento_normais:.2f}")
print(f"{horas_extras:.2f}h extras → R${pagamento_extras:.2f}")
print(f"Total: R${pagamento_total:.2f}")
```


CDB

```
# Solicita os dados do usuário
capital_inicial = float(input("Digite o valor do investimento (R$): "))
taxa_juros_percentual = float(input("Digite a taxa de juros anual (%): "))
tempo_anos = int(input("Digite o tempo do investimento (em anos): "))
```

```
# Converte a taxa percentual para decimal
taxa_juros = taxa_juros_percentual / 100
```

```
# Calcula o montante final usando a fórmula dos juros compostos
montante = capital_inicial * (1 + taxa_juros) ** tempo_anos
```

```
# Exibe o resultado formatado
print(f"\nMontante final após {tempo_anos} anos: R$ {montante:.2f}")
```

Gorjeta

```
# Solicita os dados do usuário
valor_conta = float(input("Digite o valor da conta (R$): "))
valor_gorjeta = float(input("Digite o valor da gorjeta (R$): "))
numero_pessoas = int(input("Digite o número de pessoas: "))
```

```
# Calcula a taxa de serviço (10% da conta)
taxa_servico = valor_conta * 0.10
```

```
# Calcula o valor total a ser pago (conta + taxa de serviço + gorjeta)
valor_total = valor_conta + taxa_servico + valor_gorjeta
```

```
# Calcula quanto cada pessoa deve pagar
valor_por_pessoa = valor_total / numero_pessoas
```

```
# Exibe os resultados
print(f"\nTaxa de serviço (10%): R$ {taxa_servico:.2f}")
print(f"Valor total com taxa e gorjeta: R$ {valor_total:.2f}")
print(f"Valor por pessoa: R$ {valor_por_pessoa:.2f}")
```
